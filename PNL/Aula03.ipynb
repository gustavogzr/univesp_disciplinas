```python
# Carregar o corpus
poema = """E agora, José?
A festa acabou,
a luz apagou,
o povo sumiu,
a noite esfriou,
e agora, José?
e agora, você?
você que é sem nome,
que zomba dos outros,
você que faz versos,
que ama, protesta?
e agora, José?

Está sem mulher,
está sem discurso,
está sem carinho,
já não pode beber,
já não pode fumar,
cuspir já não pode,
a noite esfriou,
o dia não veio,
o bonde não veio,
o riso não veio,
não veio a utopia
e tudo acabou
e tudo fugiu
e tudo mofou,
e agora, José?

E agora, José?
Sua doce palavra,
seu instante de febre,
sua gula e jejum,
sua biblioteca,
sua lavra de ouro,
seu terno de vidro,
sua incoerência,
seu ódio - e agora?

Com a chave na mão
quer abrir a porta,
não existe porta;
quer morrer no mar,
mas o mar secou;
quer ir para Minas,
Minas não há mais.
José, e agora?

Se você gritasse,
se você gemesse,
se você tocasse
a valsa vienense,
se você dormisse,
se você cansasse,
se você morresse...
Mas você não morre,
você é duro, José!

Sozinho no escuro
qual bicho-do-mato,
sem teogonia,
sem parede nua
para se encostar,
sem cavalo preto
que fuja a galope,
você marcha, José!
José, para onde?"""
```

```python
# Tokenização do texto, ou seja, separar as palavras
import nltk
from nltk import tokenize
nltk.download('punkt') # Baixar o tokenizador
```

```python
# Armazenar as sentenças em uma lista
poema = poema.lower().split('\n')
poema
```

```python
texto_tokenizado = []
for verso in poema:
    tokens = tokenize.word_tokenize(verso, language='portuguese') # Tokenizar o texto
    texto_tokenizado.append(tokens)

texto_tokenizado
```

```python
# Inserir os marcadore de início e fim de sentença
from nltk.lm.preprocessing import pad_both_ends

n_gramas = 2

texto_tok_pad = []

for verso in texto_tokenizado:
    verso_pad = list(pad_both_ends(verso, n=n_gramas))
    texto_tok_pad.append(verso_pad)

texto_tok_pad
```

```python
print('Lembrando que o n-gramas é igual a', n_gramas)

# gerar os bigramas
bigramadas_poema = []
for verso in texto_tok_pad:
    bigramadas = list(nltk.ngrams(verso, n = n_gramas))
    bigramadas_poema.extend(bigramadas)

bigramadas_poema
```

```python
from nltk.util import everygrams
# o everygrams gera todos os n-gramas possíveis

tam_n_gramas = 3 # define o tamanho máximo do n-grama

ngramas_poema = []
for verso in texto_tok_pad:
    ngramas = list(everygrams(verso, max_len = tam_n_gramas))
    ngramas_poema.extend(ngramas)

ngramas_poema
```

```python
from nltk.lm.preprocessing import flatten
# flatten transforma uma lista de listas em uma lista única
tokens = list(flatten(texto_tok_pad))
tokens
```

```python
# Definindo o vocabulário
from nltk.lm import Vocabulary
# Vocabulary cria um vocabulário a partir de uma lista de tokens. O vocabulário é um dicionário que mapeia palavras para índices e vice-versa.
vocab = Vocabulary(tokens, unk_cutoff=1) 
# unk_cutoff determina a partir de quantas vezes uma palavra precisa aparecer para ser considerada no vocabulário. 
vocab.counts
```

```python
# Dá para simplificar todos os passos realizados acima utilizando os comandos abaixo:
from nltk.lm.preprocessing import padded_everygram_pipeline
ngrams = 3
ngrams_pad, vocab = padded_everygram_pipeline(ngrams, texto_tokenizado)
```

```python
# Treinando o modelo de linguagem
from nltk.lm import MLE
# MLE é um modelo de linguagem que estima a probabilidade de uma palavra dado um contexto.
ngrams = 3
lm = MLE(ngrams) # Cria um modelo de linguagem MLE com n-gramas
lm.fit(ngrams_pad, vocab) # Treina o modelo de linguagem com os n-gramas e o vocabulário
```

```python
lm.generate(5, text_seed=["você"])
# lm.generate gera uma sequência de palavras a partir de um texto semente. O primeiro argumento é o número de palavras a serem geradas. O segundo argumento é o texto semente.
```

## Similaridade Textual

```python
import pandas as pd
import numpy as np
```

```python
texts = ['Goku is a hero in the Dragon Ball since 1989! Goku saved the earth so many times.',
         'The 7 Dragon balls can make wishes come true! Each ball contains his own dragon.',
         'If the wishes are superfluous, the dragon balls will become dark.' ,
         'Seiya is a bronze knight and is one of the main Knights of the Zodiac. He saved Athena several times.',
         "A knight of the zodiac wear a bronze, silver or a gold cloth to protect Athena.",
         'Saint Seiya: Knights of the Zodiac is a Japanese manga in which mystical warriors called the Saints fight wearing sacred cloths.']
classes = ['Dragon Ball', 'Dragon Ball', 'Dragon Ball', 'Cav. Zod.', 'Cav. Zod.', 'Cav. Zod.']

df = pd.DataFrame({'texts': texts, 'classes': classes})
print(df)
```

```python
# Similaridade textual baseada em caractéres

# A distância de Levenshtein é uma métrica de dissimilaridade entre duas strings. O valor da distância descreve o número mínimo de edições necessárias para transformar uma string na outra.

def levenshtein(seq1, seq2):
    # criar a matriz:
    size_x = len(seq1) + 1
    size_y = len(seq2) + 1
    matrix = np.zeros ((size_x, size_y))

    # setar número das colunas (0, n-1)
    for x in range(size_x):
        matrix [x, 0] = x
    # setar número das linhas (0, n-1)
    for y in range(size_y):
        matrix [0, y] = y

    # Calcular distância de Levenshtein
    for x in range(1, size_x): # fazer loop de 1 até n
        for y in range(1, size_y): # fazer loop de 1 até n
            if seq1[x-1] == seq2[y-1]: # se os caracteres forem iguais não aumenta a distância
                matrix [x,y] = matrix [x-1, y-1]
            else:
                matrix [x,y] = min(
                    matrix[x-1, y] + 1, # deletar
                    matrix[x-1, y-1] + 1, # substituir
                    matrix[x, y-1] + 1 # inserir
                )
    print(pd.DataFrame(matrix[1:,1:], index=list(seq1), columns=list(seq2)))

    return (matrix[size_x - 1, size_y - 1])

levenshtein('Goku', 'Gohan')
print('\n')
levenshtein('medicina', 'medico')
print('\n')
levenshtein('seiya', 'shiryu')
```

```python
# Similaridade textual baseada em termos
from sklearn.feature_extraction.text import CountVectorizer
vetorizador = CountVectorizer(
    lowercase=True, # converte o texto para minúsculo
    stop_words=['is','are','a','or','to','in','the','so','since','many','of'], # remove as stopwords
    min_df=2, # só vai considerar termos que aparecem em pelo menos 2 documentos
    dtype=np.int16 # usa inteiros de 16 bits para economizar memória
)
```

```python
# obtendo o vocabulário
vetorizador.fit(df['texts'])
print(vetorizador.get_feature_names_out())
```

```python
# Gerar representação estruturada a partir do vocabulário ajustado
representacao = vetorizador.transform(df['texts'])
# A representação retornada é uma matriz em formato esparso, ou seja, só armazena os valores diferentes de zero
representacao
```

```python
rep_array = representacao.toarray() # converte a matriz esparsa para um array
print(rep_array)
```

```python
# Similaridade por Cosseno
from sklearn.metrics.pairwise import cosine_similarity
matrix_sim = cosine_similarity(rep_array)
matrix_sim
```

```python
def most_similar(id_doc, matrix): # buscar texto mais similar
    matrix_sim = cosine_similarity(matrix) # calcular a similaridade
    best_sim = -1 # inicializa a melhor similaridade
    id_best_sim = -1 # inicializa o id do documento mais similar
    for id, doc_sim in enumerate(matrix_sim[id_doc]):
        if id != id_doc:
            if doc_sim > best_sim:
                best_sim = doc_sim
                id_best_sim = id
    return id_best_sim
```

```python
print(df)
print('\n')
doc_id = 5
print('Texto:', df.loc[doc_id, 'texts'])
print('Texto mais parecido:', df.loc[most_similar(doc_id, rep_array), 'texts'])
```

```python
# Similaridade de Jaccard
# É calculada pela divisão do número de elementos em comum entre dois conjuntos pelo número total de elementos nos conjuntos.
from sklearn.metrics.pairwise import pairwise_distances

def most_similar(id_doc, matrix):
    matrix_sim = 1 - pairwise_distances(matrix, metric='jaccard') # calcular a similaridade
    best_sim = -1 # inicializa a melhor similaridade
    id_best_sim = -1 # inicializa o id do documento mais similar
    for id, doc_sim in enumerate(matrix_sim[id_doc]):
        if id != id_doc:
            if doc_sim > best_sim:
                best_sim = doc_sim
                id_best_sim = id
    return id_best_sim
```
